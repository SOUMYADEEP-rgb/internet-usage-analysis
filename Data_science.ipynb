```python
# Load column names
with open("col.txt", "r") as f:
    columns = [line.strip() for line in f.readlines()]

print("Total columns:", len(columns))
print("First 10 columns:", columns[:10])
```

```python
rows = []

with open("data.txt", "r") as f:
    for line in f:
        # Split manually on whitespace
        values = line.strip().split()
        rows.append(values)

print("Total rows:", len(rows))
print("First row length:", len(rows[0]))
print("First row preview:", rows[0][:10])
```

```python
# Keep only correct rows
clean_rows = [row for row in rows if len(row) == 72]

print("Rows kept:", len(clean_rows))
```

```python
data_full = pd.DataFrame(clean_rows)
data_full = data_full.apply(pd.to_numeric, errors="coerce")

print("Clean dataset shape:", data_full.shape)
```

```python
selected_cols = [
    "Age",
    "Education Attainment",
    "Household Income",
    "Major Occupation",
    "Years on Internet",
    "Web Ordering",
    "Web Page Creation",
    "Willingness to Pay Fees",
    "Primary Place of WWW Access",
    "Who Pays for Access_Self",
    "Who Pays for Access_Work",
    "Community Building"
]

data_full[selected_cols].info()
```

```python
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA

X = data_full[selected_cols]

# Standardize
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# PCA
pca = PCA()
X_pca = pca.fit_transform(X_scaled)

# Explained variance
print("Explained variance ratio:")
print(pca.explained_variance_ratio_)

# Use PC1 as index
data_full["Digital_Readiness_Index"] = X_pca[:, 0]

data_full[["Digital_Readiness_Index"]].head()
```

```python
import numpy as np

cumulative_variance = np.cumsum(pca.explained_variance_ratio_)
print(cumulative_variance)
```

```python
import numpy as np

# Number of components to use
k = 6

# Get first 6 components
components = X_pca[:, :k]

# Get their explained variance ratios
weights = pca.explained_variance_ratio_[:k]

# Normalize weights (optional but cleaner)
weights = weights / weights.sum()

# Construct weighted index
digital_index = np.dot(components, weights)

# Add to dataframe
data_full["Digital_Readiness_Index"] = digital_index

data_full["Digital_Readiness_Index"].describe()
```

```python
from sklearn.preprocessing import MinMaxScaler

scaler = MinMaxScaler()
data_full["Digital_Readiness_Index"] = scaler.fit_transform(
    data_full[["Digital_Readiness_Index"]]
) * 100

data_full["Digital_Readiness_Index"].describe()
```
